```python
import pandas as pd

# Đọc file parquet đã xử lý
parquet_path = "../data/processed_v1/user_010.parquet"
df = pd.read_parquet(parquet_path)

# Lọc một đoạn ngắn có chuyển đổi giữa 'unknown', 'taxi', 'walk' để xem trực quan
df_sample = df[
    (df['datetime'] >= '2008-04-01 00:48:00') & 
    (df['datetime'] <= '2008-04-01 01:05:00')
]

# 1. Hiển thị bảng mặc định
display(df_sample.head(20))

# 2. Định dạng màu sắc trực quan (Highlight theo cột mode & speed)
styled_df = df_sample.head(20).style \
    .format({'speed_kmh': '{:.1f} km/h', 'altitude_m': '{:.1f} m'}) \
    .background_gradient(subset=['speed_kmh'], cmap='YlOrRd') \
    .map(lambda val: 'font-weight: bold; color: #2E7D32;' if val != 'unknown' else 'color: #757575;', subset=['mode'])

display(styled_df)
```
